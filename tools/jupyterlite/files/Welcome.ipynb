## A quick data analysis

````python
df = pd.DataFrame({
    'subject': ['DSA', 'DBMS', 'OS', 'CN', 'AI'],
    'marks': [86, 78, 91, 72, 88],
})
df['grade'] = pd.cut(df['marks'], [0, 60, 75, 90, 100], labels=['C', 'B', 'A', 'O'])
df.describe()
````

````python
import matplotlib.pyplot as plt

ax = df.plot.bar(x='subject', y='marks', legend=False, color='#342e6c')
ax.set_ylabel('Marks')
ax.set_title('Marks by subject')
plt.show()
````

## Machine learning in a few lines

````python
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier

X, y = load_iris(return_X_y=True)
X_train, X_test, y_train, y_test = train_test_split(X, y, random_state=0)
model = KNeighborsClassifier(n_neighbors=5).fit(X_train, y_train)
print('Test accuracy:', round(model.score(X_test, y_test), 3))
````
